# ABI-DAGAR run for the matched-MCMC benchmark

This notebook applies the frozen ABI posterior operator to the fixed held-out bank used by the prior- and construction-matched MCMC experiment. It saves parameter summaries, edge-level boundary probabilities, runtime, and compressed posterior draws.

```python
from pathlib import Path
import os
import sys

from IPython.display import display

def find_repository_root(start=None):
    current = (Path.cwd() if start is None else Path(start)).resolve()
    for candidate in (current, *current.parents):
        marker = candidate / 'Simulation Experiments' / 'ABI_vs_matched_MCMC' / 'matched_abi_runner.py'
        if marker.exists():
            return candidate
    raise FileNotFoundError('Could not locate the ABI_poisson_regression repository root.')

REPOSITORY_ROOT = find_repository_root()
EXPERIMENT_DIR = REPOSITORY_ROOT / 'Simulation Experiments' / 'ABI_vs_matched_MCMC'
if str(EXPERIMENT_DIR) not in sys.path:
    sys.path.insert(0, str(EXPERIMENT_DIR))

from matched_abi_runner import run_abi_benchmark

BENCHMARK_DIR = EXPERIMENT_DIR / 'datasets' / 'benchmark_bank_seed123_n100'
RESULTS_DIR = Path(os.environ.get(
    'ABI_MATCHED_ABI_RESULTS_DIR', BENCHMARK_DIR / 'abi_results_all100'
))
CHECKPOINT = Path(os.environ.get(
    'ABI_MATCHED_CHECKPOINT', REPOSITORY_ROOT / 'Training' / 'Checkpoints' / 'poisson_dagar.keras'
))
NUM_SAMPLES = int(os.environ.get('ABI_MATCHED_NUM_SAMPLES', '10000'))
SEED = 123
DATASET_IDS = None
MAX_DATASETS_ENV = os.environ.get('ABI_MATCHED_MAX_DATASETS')
MAX_DATASETS = int(MAX_DATASETS_ENV) if MAX_DATASETS_ENV else None
SAVE_DRAWS = True
QUIET = False

BENCHMARK_DIR, RESULTS_DIR, CHECKPOINT
```

## Run the frozen posterior operator

```python
outputs = run_abi_benchmark(
    benchmark_dir=BENCHMARK_DIR,
    results_dir=RESULTS_DIR,
    checkpoint=CHECKPOINT,
    num_samples=NUM_SAMPLES,
    seed=SEED,
    dataset_ids=DATASET_IDS,
    max_datasets=MAX_DATASETS,
    save_draws=SAVE_DRAWS,
    quiet=QUIET,
)

display(outputs['parameter_df'].head())
display(outputs['edge_metrics_df'].head())
display(outputs['runtime_df'].head())
```
